```python
import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# Exercise 1: Monthly Sales Trend
# ==========================================
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [120, 150, 170, 160, 190]

plt.figure(figsize=(6, 4))
plt.plot(
    months, 
    sales, 
    color='green', 
    linestyle='--', 
    marker='s', 
    markersize=8, 
    label='Sales'
)
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

```python
# ==========================================
# Exercise 2: Student Scores
# ==========================================
subjects = ['Math', 'Physics', 'Chemistry', 'Biology']
scores = [85, 78, 92, 88]

plt.figure(figsize=(6, 4))
plt.plot(
    subjects, 
    scores, 
    color='red', 
    marker='*', 
    linewidth=3, 
    markerfacecolor='yellow', 
    label='Scores'
)
plt.title('Exam Scores')
plt.xlabel('Subjects')
plt.ylabel('Scores')
plt.legend()
plt.show()

```

```python
# ==========================================
# Exercise 3: Company Revenue
# ==========================================
years = [2021, 2022, 2023, 2024]
revenue = [500, 700, 850, 1000]

plt.figure(figsize=(6, 4))
plt.plot(
    years, 
    revenue, 
    color='blue', 
    linestyle=':', 
    marker='D', 
    label='Revenue'
)
plt.title('Company Revenue Growth')
plt.xlabel('Year')
plt.ylabel('Revenue ($)')
plt.legend()
plt.show()

```

```python
# ==========================================
# Exercise 4: Fruit Supply (Vertical Bar Chart)
# ==========================================
fruits = ['Apple', 'Orange', 'Banana', 'Mango']
counts = [50, 70, 40, 60]
colors = ['red', 'orange', 'yellow', 'green']

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(fruits, counts, color=colors, label='Fruit Quantity')
ax.set_title('Fruit Supply')
ax.set_ylabel('Quantity')
ax.legend()
plt.show()

```

```python
# ==========================================
# Exercise 5: Stacked Bar Chart (Students by Gender)
# ==========================================
departments = ('CS', 'IS', 'SE', 'IT')
students = {
    'Male': np.array([70, 65, 80, 50]),
    'Female': np.array([60, 55, 70, 45])
}

width = 0.7

fig, ax = plt.subplots(figsize=(7, 5))

# Plot Male bars
p1 = ax.bar(departments, students['Male'], width=width, label='Male', color='skyblue')

# Plot Female bars stacked on top of Male bars
p2 = ax.bar(departments, students['Female'], width=width, bottom=students['Male'], label='Female', color='pink')

# Add labels inside each section
ax.bar_label(p1, label_type='center')
ax.bar_label(p2, label_type='center')

ax.set_title('Students by Department and Gender')
ax.set_ylabel('Number of Students')
ax.legend()
plt.show()
```

```python
# ==========================================
# Bonus Challenge: Build a Dashboard (4 Subplots)
# ==========================================
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# 1. Monthly Sales Trend (Line Chart)
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [120, 150, 170, 160, 190]
axes[0, 0].plot(months, sales, color='green', linestyle='--', marker='s', markersize=8, label='Sales')
axes[0, 0].set_title('Monthly Sales Trend')
axes[0, 0].set_ylabel('Sales Amount')
axes[0, 0].legend()

# 2. Fruit Supply (Vertical Bar Chart)
fruits = ['Apple', 'Orange', 'Banana', 'Mango']
counts = [50, 70, 40, 60]
fruit_colors = ['red', 'orange', 'yellow', 'green']
axes[0, 1].bar(fruits, counts, color=fruit_colors)
axes[0, 1].set_title('Fruit Supply')
axes[0, 1].set_ylabel('Quantity')

# 3. Movie Ratings (Horizontal Bar Chart)
movies = ['Inception', 'Interstellar', 'Dark Knight', 'Avatar']
ratings = [8.8, 8.7, 9.0, 7.9]
axes[1, 0].barh(movies, ratings, color='purple')
axes[1, 0].set_title('Movie Ratings')
axes[1, 0].set_xlabel('Rating')

# 4. Student Distribution by Gender (Stacked Bar Chart)
departments = ('CS', 'IS', 'SE', 'IT')
male_students = np.array([70, 65, 80, 50])
female_students = np.array([60, 55, 70, 45])

p1 = axes[1, 1].bar(departments, male_students, width=0.7, label='Male', color='skyblue')
p2 = axes[1, 1].bar(departments, female_students, width=0.7, bottom=male_students, label='Female', color='pink')

axes[1, 1].bar_label(p1, label_type='center')
axes[1, 1].bar_label(p2, label_type='center')
axes[1, 1].set_title('Student Distribution by Gender')
axes[1, 1].set_ylabel('Count')
axes[1, 1].legend()

# Adjust spacing and display
plt.tight_layout()
plt.show()
```
